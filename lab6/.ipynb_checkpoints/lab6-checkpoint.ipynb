```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.impute import SimpleImputer
import warnings

warnings.filterwarnings('ignore')

# Загрузка датасета
print("Загрузка датасета...")
data = pd.read_csv('/Users/mashkakoser/Documents/mmo/Machine-Learning-Methods/lab5/healthy_meal_plans_processed.csv')

print(f"Размер датасета: {data.shape}")
print(f"Количество признаков: {data.shape[1]}")
print(f"Количество объектов: {data.shape[0]}")

# Информация о типах данных и пропусках
print("\nИнформация о датасете:")
print(data.info())
```

```python
### 2. Анализ пропущенных значений
# Анализ пропущенных значений
print("Анализ пропущенных значений:")
print("="*50)

# Подсчет пропущенных значений по столбцам
missing_values = data.isnull().sum()
print("Пропущенные значения по столбцам:")
print(missing_values)

# Процент пропущенных значений
missing_percent = (missing_values / len(data)) * 100
print(f"\nПроцент пропущенных значений:")
print(missing_percent.round(2))

# Визуализация пропущенных значений
plt.figure(figsize=(12, 6))
sns.heatmap(data.isnull(), cbar=False, cmap='viridis')
plt.title('Тепловая карта пропущенных значений')
plt.show()

# Проверка на полностью пустые столбцы
empty_columns = missing_values[missing_values == len(data)].index.tolist()
if empty_columns:
    print(f"\nПолностью пустые столбцы: {empty_columns}")
else:
    print("\nПолностью пустых столбцов не найдено")

print(f"\nОбщее количество пропущенных значений: {missing_values.sum()}")

# Создаем копию данных для экспериментов
data_with_missing = data.copy()
```

```python
### 3. Методы заполнения пропусков
# Для демонстрации создадим искусственные пропуски в данных
print("Создание искусственных пропусков для демонстрации:")
print("="*50)

# Создаем копию данных и добавляем пропуски
data_demo = data.copy()

# Добавляем случайные пропуски в числовые столбцы
numeric_cols = ['calories', 'protein', 'fat', 'carbs']
for col in numeric_cols:
    # Добавляем 5% пропусков
    mask = np.random.random(len(data_demo)) < 0.05
    data_demo.loc[mask, col] = np.nan

print(f"Добавлено пропусков в датасет: {data_demo.isnull().sum().sum()}")

# Метод 1: Заполнение средним значением
print("\nМетод 1: Заполнение средним значением")
data_mean = data_demo.copy()
for col in numeric_cols:
    mean_val = data_mean[col].mean()
    data_mean[col].fillna(mean_val, inplace=True)
    print(f"{col}: заполнено средним = {mean_val:.3f}")

print(f"Пропусков после заполнения: {data_mean.isnull().sum().sum()}")

# Метод 2: Заполнение медианой
print("\nМетод 2: Заполнение медианой")
data_median = data_demo.copy()
for col in numeric_cols:
    median_val = data_median[col].median()
    data_median[col].fillna(median_val, inplace=True)
    print(f"{col}: заполнено медианой = {median_val:.3f}")

print(f"Пропусков после заполнения: {data_median.isnull().sum().sum()}")

# Метод 3: Заполнение константой
print("\nМетод 3: Заполнение константой")
data_constant = data_demo.copy()
for col in numeric_cols:
    data_constant[col].fillna(-999, inplace=True)
    print(f"{col}: заполнено константой -999")

print(f"Пропусков после заполнения: {data_constant.isnull().sum().sum()}")

# Метод 4: Удаление строк с пропусками
print("\nМетод 4: Удаление строк с пропусками")
data_drop = data_demo.dropna()
print(f"Исходный размер: {data_demo.shape}")
print(f"Размер после удаления: {data_drop.shape}")
print(f"Удалено строк: {data_demo.shape[0] - data_drop.shape[0]}")
```

```python
### 4. Сравнение методов и анализ результатов
# Сравнение методов заполнения пропусков
print("Сравнение методов заполнения пропусков:")
print("="*50)

# Сравним статистику до и после заполнения для одного столбца
col = 'calories'
print(f"\nСтатистика для столбца '{col}':")
print(f"Исходные данные: среднее = {data[col].mean():.3f}, медиана = {data[col].median():.3f}")
print(f"После заполнения средним: среднее = {data_mean[col].mean():.3f}")
print(f"После заполнения медианой: медиана = {data_median[col].median():.3f}")
print(f"После удаления строк: размер = {data_drop.shape[0]}")

# Визуализация сравнения методов
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.hist(data[col], bins=30, alpha=0.7, label='Исходные')
plt.hist(data_mean[col], bins=30, alpha=0.7, label='Среднее')
plt.title('Сравнение распределений')
plt.legend()

plt.subplot(1, 3, 2)
plt.hist(data[col], bins=30, alpha=0.7, label='Исходные')
plt.hist(data_median[col], bins=30, alpha=0.7, label='Медиана')
plt.title('Сравнение с медианой')
plt.legend()

plt.subplot(1, 3, 3)
plt.hist(data[col], bins=30, alpha=0.7, label='Исходные')
plt.hist(data_constant[col], bins=30, alpha=0.7, label='Константа')
plt.title('Сравнение с константой')
plt.legend()

plt.tight_layout()
plt.show()

print("\nАнализ методов:")
print("="*50)

# Подсчет количества пропусков в исходном датасете
original_missing = data.isnull().sum().sum()
print(f"Исходное количество пропусков: {original_missing}")

if original_missing > 0:
    print("РЕЗУЛЬТАТЫ ОБРАБОТКИ ПРОПУСКОВ:")    
    print("✅ Датасет проанализирован на пропуски")
    print("✅ Протестированы 4 метода заполнения:")
    print("  • Заполнение средним значением")
    print("  • Заполнение медианой")
    print("  • Заполнение константой")
    print("  • Удаление строк с пропусками")
    print("✅ Данные готовы для дальнейшего анализа")
else:
    print("ОТЛИЧНЫЕ НОВОСТИ!")
    print("✅ В датасете НЕТ пропущенных значений!")
    print("✅ Данные уже чистые и готовы к использованию")
    print("✅ Можно переходить к следующему этапу анализа")

print(f"\nРазмер датасета: {data.shape}")
print(f"Количество признаков: {data.shape[1]}")
print(f"Количество объектов: {data.shape[0]}")
```
